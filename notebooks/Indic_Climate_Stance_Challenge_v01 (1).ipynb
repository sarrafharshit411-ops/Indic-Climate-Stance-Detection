```python
# ============================================================
# CSET346 — INDIC CLIMATE STANCE CHALLENGE
# PHASE 1: COMPLETE DATASET EDA
# ============================================================

# =========================
# 1. INSTALL REQUIREMENTS
# =========================
!pip -q install pandas numpy matplotlib seaborn wordcloud openpyxl

# =========================
# 2. IMPORT LIBRARIES
# =========================
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from collections import Counter
from wordcloud import WordCloud
from google.colab import files

print("Libraries loaded successfully.")

# =========================
# 3. CREATE RESULT FOLDERS
# =========================
os.makedirs("results/tables", exist_ok=True)
os.makedirs("results/figures", exist_ok=True)

# =========================
# 4. UPLOAD DATASETS
# =========================
print("Upload these 3 files:")
print("1. train(1).csv")
print("2. val(1).csv")
print("3. test(1).csv")

uploaded = files.upload()

print("\nUploaded files:")
for f in uploaded:
    print(" -", f)

# =========================
# 5. FIND CSV FILES
# =========================
csv_files = [f for f in uploaded.keys() if f.lower().endswith(".csv")]

if len(csv_files) < 3:
    raise ValueError("Please upload train, validation and test CSV files.")

def find_file(keyword):
    for f in csv_files:
        if keyword.lower() in f.lower():
            return f
    return None

train_file = find_file("train")
val_file = find_file("val")
test_file = find_file("test")

if train_file is None or val_file is None or test_file is None:
    print("\nCould not automatically identify all files.")
    print("Available files:", csv_files)
    raise ValueError("Rename files clearly as train, val and test and rerun.")

print("\nFiles selected:")
print("Train:", train_file)
print("Validation:", val_file)
print("Test:", test_file)

# =========================
# 6. LOAD DATA
# =========================
train = pd.read_csv(train_file)
val = pd.read_csv(val_file)
test = pd.read_csv(test_file)

print("\n================ DATASET SHAPES ================")
print("Train      :", train.shape)
print("Validation :", val.shape)
print("Test       :", test.shape)

# =========================
# 7. BASIC STRUCTURE
# =========================
print("\n================ COLUMNS ================")
print("Train columns:", train.columns.tolist())
print("Validation columns:", val.columns.tolist())
print("Test columns:", test.columns.tolist())

required_columns = {"text", "label"}

for name, df in [
    ("Train", train),
    ("Validation", val),
    ("Test", test)
]:
    missing = required_columns - set(df.columns)

    if missing:
        raise ValueError(
            f"{name} is missing columns: {missing}"
        )

print("\nRequired columns found: text, label")

# =========================
# 8. LABEL MAPPING
# =========================
# Dataset-derived mapping established from inspection
label_map = {
    0: "Against",
    1: "Favour",
    2: "None"
}

for df in [train, val, test]:
    df["label_name"] = df["label"].map(label_map)

unknown_labels = set(
    pd.concat([train["label"], val["label"], test["label"]]).unique()
) - set(label_map.keys())

if unknown_labels:
    print("WARNING: Unknown labels found:", unknown_labels)

print("\n================ LABEL MAPPING ================")
for k, v in label_map.items():
    print(f"{k} → {v}")

# =========================
# 9. MISSING VALUES
# =========================
print("\n================ MISSING VALUES ================")

for name, df in [
    ("Train", train),
    ("Validation", val),
    ("Test", test)
]:
    print(f"\n{name}")
    print(df[["text", "label"]].isnull().sum())

# =========================
# 10. DUPLICATES
# =========================
print("\n================ DUPLICATES ================")

for name, df in [
    ("Train", train),
    ("Validation", val),
    ("Test", test)
]:
    print(f"{name}:")
    print("Duplicate rows :", df.duplicated().sum())
    print("Duplicate texts:", df["text"].duplicated().sum())

# =========================
# 11. LABEL DISTRIBUTION
# =========================
print("\n================ LABEL DISTRIBUTION ================")

def label_distribution(df, name):
    counts = df["label"].value_counts().sort_index()
    percentages = counts / len(df) * 100

    result = pd.DataFrame({
        "Label": counts.index,
        "Class": [label_map.get(x, "Unknown") for x in counts.index],
        "Count": counts.values,
        "Percentage": percentages.values
    })

    print("\n", name)
    print(result.to_string(index=False))

    return result

train_dist = label_distribution(train, "TRAIN")
val_dist = label_distribution(val, "VALIDATION")
test_dist = label_distribution(test, "TEST")

train_dist.to_csv(
    "results/tables/train_label_distribution.csv",
    index=False
)

val_dist.to_csv(
    "results/tables/validation_label_distribution.csv",
    index=False
)

test_dist.to_csv(
    "results/tables/test_label_distribution.csv",
    index=False
)

# =========================
# 12. COMBINED DATASET
# =========================
train["split"] = "Train"
val["split"] = "Validation"
test["split"] = "Test"

all_data = pd.concat(
    [train, val, test],
    ignore_index=True
)

print("\n================ TOTAL DATA ================")
print("Total samples:", len(all_data))
print("Train:", len(train))
print("Validation:", len(val))
print("Test:", len(test))

# =========================
# 13. TEXT LENGTH FEATURES
# =========================

def add_text_features(df):

    df["text"] = df["text"].astype(str)

    df["char_count"] = df["text"].str.len()

    df["word_count"] = df["text"].apply(
        lambda x: len(x.split())
    )

    df["sentence_count"] = df["text"].apply(
        lambda x: max(1, len(re.findall(r"[.!?]+", x)))
    )

    df["url_count"] = df["text"].apply(
        lambda x: len(
            re.findall(r"https?://\S+|www\.\S+", x)
        )
    )

    df["mention_count"] = df["text"].apply(
        lambda x: len(re.findall(r"@\w+", x))
    )

    df["hashtag_count"] = df["text"].apply(
        lambda x: len(re.findall(r"#\w+", x))
    )

    df["exclamation_count"] = df["text"].str.count("!")

    df["question_count"] = df["text"].str.count(r"\?")

    df["uppercase_count"] = df["text"].apply(
        lambda x: sum(c.isupper() for c in x)
    )

    df["digit_count"] = df["text"].apply(
        lambda x: sum(c.isdigit() for c in x)
    )

    df["non_ascii_count"] = df["text"].apply(
        lambda x: sum(ord(c) > 127 for c in x)
    )

    df["emoji_like"] = df["text"].apply(
        lambda x: bool(
            re.search(
                r"[\U0001F300-\U0001FAFF]",
                x
            )
        )
    )

    return df

all_data = add_text_features(all_data)

# =========================
# 14. TEXT STATISTICS
# =========================
print("\n================ TEXT STATISTICS ================")

text_stats = all_data[
    [
        "char_count",
        "word_count",
        "sentence_count",
        "url_count",
        "mention_count",
        "hashtag_count",
        "exclamation_count",
        "question_count",
        "uppercase_count",
        "digit_count",
        "non_ascii_count"
    ]
].describe().T

print(text_stats)

text_stats.to_csv(
    "results/tables/text_statistics.csv"
)

# =========================
# 15. STATISTICS BY SPLIT
# =========================
print("\n================ STATISTICS BY SPLIT ================")

split_stats = all_data.groupby("split")[
    [
        "char_count",
        "word_count",
        "url_count",
        "mention_count",
        "hashtag_count",
        "non_ascii_count"
    ]
].mean().round(2)

print(split_stats)

split_stats.to_csv(
    "results/tables/split_statistics.csv"
)

# =========================
# 16. STATISTICS BY CLASS
# =========================
print("\n================ STATISTICS BY CLASS ================")

class_stats = all_data.groupby("label_name")[
    [
        "char_count",
        "word_count",
        "url_count",
        "mention_count",
        "hashtag_count",
        "non_ascii_count"
    ]
].mean().round(2)

print(class_stats)

class_stats.to_csv(
    "results/tables/class_statistics.csv"
)

# =========================
# 17. CLASS DISTRIBUTION GRAPH
# =========================
plt.figure(figsize=(8, 5))

sns.countplot(
    data=all_data,
    x="label_name",
    order=["Against", "Favour", "None"]
)

plt.title("Overall Climate Stance Distribution")
plt.xlabel("Stance")
plt.ylabel("Number of Samples")
plt.tight_layout()

plt.savefig(
    "results/figures/class_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# 18. CLASS DISTRIBUTION BY SPLIT
# =========================
plt.figure(figsize=(9, 5))

sns.countplot(
    data=all_data,
    x="split",
    hue="label_name",
    hue_order=["Against", "Favour", "None"]
)

plt.title("Class Distribution Across Dataset Splits")
plt.xlabel("Dataset Split")
plt.ylabel("Number of Samples")
plt.tight_layout()

plt.savefig(
    "results/figures/class_distribution_by_split.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# 19. WORD COUNT DISTRIBUTION
# =========================
plt.figure(figsize=(10, 6))

sns.histplot(
    data=all_data,
    x="word_count",
    bins=40,
    kde=True
)

plt.title("Distribution of Word Counts")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig(
    "results/figures/word_count_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# 20. WORD COUNT BY CLASS
# =========================
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=all_data,
    x="label_name",
    y="word_count",
    order=["Against", "Favour", "None"]
)

plt.title("Text Length by Stance Class")
plt.xlabel("Stance")
plt.ylabel("Word Count")
plt.tight_layout()

plt.savefig(
    "results/figures/word_count_by_class.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# 21. CHARACTER COUNT
# =========================
plt.figure(figsize=(10, 6))

sns.histplot(
    data=all_data,
    x="char_count",
    bins=50,
    kde=True
)

plt.title("Distribution of Character Counts")
plt.xlabel("Number of Characters")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig(
    "results/figures/character_count_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# 22. SOCIAL MEDIA FEATURES
# =========================
feature_means = all_data.groupby("label_name")[
    [
        "url_count",
        "mention_count",
        "hashtag_count",
        "exclamation_count",
        "question_count"
    ]
].mean()

print("\n================ SOCIAL MEDIA FEATURES BY CLASS ================")
print(feature_means.round(3))

feature_means.to_csv(
    "results/tables/social_media_features_by_class.csv"
)

# =========================
# 23. SOCIAL MEDIA FEATURE GRAPH
# =========================
plot_df = feature_means.reset_index()

plot_df = plot_df.melt(
    id_vars="label_name",
    var_name="Feature",
    value_name="Average"
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=plot_df,
    x="Feature",
    y="Average",
    hue="label_name"
)

plt.title("Social-Media Features by Stance Class")
plt.xlabel("Feature")
plt.ylabel("Average Count")
plt.xticks(rotation=30)
plt.tight_layout()

plt.savefig(
    "results/figures/social_media_features.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# 24. NON-ASCII / POSSIBLE MULTILINGUAL TEXT
# =========================
non_ascii_summary = all_data.groupby("label_name")[
    "non_ascii_count"
].mean()

print("\n================ NON-ASCII ANALYSIS ================")
print(non_ascii_summary.round(3))

# =========================
# 25. URL / HASHTAG / MENTION PRESENCE
# =========================
all_data["has_url"] = all_data["url_count"] > 0
all_data["has_mention"] = all_data["mention_count"] > 0
all_data["has_hashtag"] = all_data["hashtag_count"] > 0

presence_table = pd.DataFrame({
    "Has URL (%)":
        all_data.groupby("label_name")["has_url"].mean() * 100,

    "Has Mention (%)":
        all_data.groupby("label_name")["has_mention"].mean() * 100,

    "Has Hashtag (%)":
        all_data.groupby("label_name")["has_hashtag"].mean() * 100
}).round(2)

print("\n================ SOCIAL MEDIA PRESENCE ================")
print(presence_table)

presence_table.to_csv(
    "results/tables/social_media_presence.csv"
)

# =========================
# 26. MOST COMMON WORDS
# =========================

def clean_text(text):

    text = text.lower()

    # remove URLs
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )

    # remove mentions
    text = re.sub(
        r"@\w+",
        " ",
        text
    )

    # remove hashtags symbol but retain word
    text = re.sub(
        r"#",
        "",
        text
    )

    # keep alphabetic words
    words = re.findall(
        r"[a-zA-Z]{2,}",
        text
    )

    return words


stopwords = {
    "the","and","for","that","this","with","are","was",
    "from","have","has","had","you","your","they","their",
    "will","would","could","should","about","there","what",
    "when","where","which","who","why","how","not","but",
    "our","out","all","can","its","into","than","then",
    "them","his","her","she","him","been","were","also",
    "more","one","two","very","just","only","over","get",
    "got","like","people","know","use","using"
}

def common_words(df, n=30):

    counter = Counter()

    for text in df["text"]:
        words = clean_text(str(text))

        words = [
            w for w in words
            if w not in stopwords
        ]

        counter.update(words)

    return counter.most_common(n)

overall_words = common_words(all_data, 50)

common_words_df = pd.DataFrame(
    overall_words,
    columns=["Word", "Frequency"]
)

print("\n================ MOST COMMON WORDS ================")
print(common_words_df.to_string(index=False))

common_words_df.to_csv(
    "results/tables/most_common_words.csv",
    index=False
)

# =========================
# 27. WORDCLOUD
# =========================
text_for_wc = " ".join(
    all_data["text"].astype(str).tolist()
)

wordcloud = WordCloud(
    width=1400,
    height=700,
    background_color="white",
    stopwords=stopwords,
    collocations=False
).generate(text_for_wc)

plt.figure(figsize=(15, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Climate Stance Dataset Word Cloud")
plt.tight_layout()

plt.savefig(
    "results/figures/wordcloud.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================
# 28. CLASS-SPECIFIC WORDS
# =========================
for class_name in ["Against", "Favour", "None"]:

    subset = all_data[
        all_data["label_name"] == class_name
    ]

    words = common_words(subset, 30)

    df_words = pd.DataFrame(
        words,
        columns=["Word", "Frequency"]
    )

    print(f"\n========== {class_name.upper()} WORDS ==========")
    print(df_words.to_string(index=False))

    safe_name = class_name.lower()

    df_words.to_csv(
        f"results/tables/{safe_name}_common_words.csv",
        index=False
    )

# =========================
# 29. LEAKAGE CHECK
# =========================
print("\n================ DATA LEAKAGE CHECK ================")

train_texts = set(train["text"].astype(str))
val_texts = set(val["text"].astype(str))
test_texts = set(test["text"].astype(str))

train_val_overlap = train_texts.intersection(val_texts)
train_test_overlap = train_texts.intersection(test_texts)
val_test_overlap = val_texts.intersection(test_texts)

print(
    "Train ↔ Validation:",
    len(train_val_overlap)
)

print(
    "Train ↔ Test:",
    len(train_test_overlap)
)

print(
    "Validation ↔ Test:",
    len(val_test_overlap)
)

leakage_summary = pd.DataFrame({
    "Comparison": [
        "Train-Validation",
        "Train-Test",
        "Validation-Test"
    ],
    "Exact_Text_Overlap": [
        len(train_val_overlap),
        len(train_test_overlap),
        len(val_test_overlap)
    ]
})

leakage_summary.to_csv(
    "results/tables/leakage_check.csv",
    index=False
)

# =========================
# 30. RANDOM EXAMPLES
# =========================
print("\n================ RANDOM EXAMPLES BY CLASS ================")

for class_name in ["Against", "Favour", "None"]:

    print(f"\n\n========== {class_name.upper()} ==========")

    subset = all_data[
        all_data["label_name"] == class_name
    ]

    examples = subset.sample(
        min(5, len(subset)),
        random_state=42
    )

    for i, row in enumerate(
        examples["text"],
        start=1
    ):
        print(f"\n{i}. {row}")

# =========================
# 31. VERY LONG TEXTS
# =========================
print("\n================ LONGEST TEXTS ================")

longest = all_data.nlargest(
    10,
    "word_count"
)[
    ["label_name", "word_count", "text"]
]

print(longest.to_string(index=False))

longest.to_csv(
    "results/tables/longest_texts.csv",
    index=False
)

# =========================
# 32. VERY SHORT TEXTS
# =========================
print("\n================ SHORTEST TEXTS ================")

shortest = all_data.nsmallest(
    10,
    "word_count"
)[
    ["label_name", "word_count", "text"]
]

print(shortest.to_string(index=False))

shortest.to_csv(
    "results/tables/shortest_texts.csv",
    index=False
)

# =========================
# 33. FINAL EDA SUMMARY
# =========================

summary = {
    "Total Samples": len(all_data),
    "Training Samples": len(train),
    "Validation Samples": len(val),
    "Test Samples": len(test),

    "Against Count": int(
        (all_data["label_name"] == "Against").sum()
    ),

    "Favour Count": int(
        (all_data["label_name"] == "Favour").sum()
    ),

    "None Count": int(
        (all_data["label_name"] == "None").sum()
    ),

    "Average Words": round(
        all_data["word_count"].mean(), 2
    ),

    "Median Words": round(
        all_data["word_count"].median(), 2
    ),

    "Average Characters": round(
        all_data["char_count"].mean(), 2
    ),

    "URLs (%)": round(
        all_data["has_url"].mean() * 100, 2
    ),

    "Mentions (%)": round(
        all_data["has_mention"].mean() * 100, 2
    ),

    "Hashtags (%)": round(
        all_data["has_hashtag"].mean() * 100, 2
    ),

    "Non-ASCII (%)": round(
        (all_data["non_ascii_count"] > 0).mean() * 100,
        2
    ),

    "Train-Val Exact Overlap":
        len(train_val_overlap),

    "Train-Test Exact Overlap":
        len(train_test_overlap),

    "Val-Test Exact Overlap":
        len(val_test_overlap)
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=["Metric", "Value"]
)

print("\n\n================================================")
print("              FINAL EDA SUMMARY")
print("================================================")
print(summary_df.to_string(index=False))

summary_df.to_csv(
    "results/tables/final_eda_summary.csv",
    index=False
)

# =========================
# 34. SAVE PROCESSED DATA
# =========================
all_data.to_csv(
    "results/tables/combined_eda_data.csv",
    index=False
)

# =========================
# 35. ZIP ALL RESULTS
# =========================
import shutil

shutil.make_archive(
    "CSET346_EDA_Results",
    "zip",
    "results"
)

print("\n================================================")
print("EDA COMPLETE!")
print("================================================")

print("\nGenerated:")
print("✓ Dataset statistics")
print("✓ Label distributions")
print("✓ Text-length analysis")
print("✓ URL analysis")
print("✓ Mention analysis")
print("✓ Hashtag analysis")
print("✓ Non-ASCII analysis")
print("✓ Common-word analysis")
print("✓ Word clouds")
print("✓ Class-specific vocabulary")
print("✓ Duplicate analysis")
print("✓ Leakage analysis")
print("✓ Random examples")
print("✓ Long/short text analysis")
print("✓ CSV tables")
print("✓ PNG figures")

print("\nResults folder: results/")
print("ZIP file: CSET346_EDA_Results.zip")

# =========================
# 36. DOWNLOAD RESULTS
# =========================
files.download("CSET346_EDA_Results.zip")
```

```python
# ============================================================
# CSET346 — BASELINE 1
# TF-IDF + LOGISTIC REGRESSION
# ============================================================

!pip -q install pandas numpy scikit-learn matplotlib seaborn

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    f1_score,
    classification_report,
    confusion_matrix
)

print("Libraries loaded successfully.")

# ============================================================
# 1. UPLOAD DATA
# ============================================================

print("Upload:")
print("train(1).csv")
print("val(1).csv")
print("test(1).csv")

uploaded = files.upload()

csv_files = list(uploaded.keys())

def find_file(keyword):
    for f in csv_files:
        if keyword.lower() in f.lower():
            return f
    return None

train_file = find_file("train")
val_file = find_file("val")
test_file = find_file("test")

print("\nFiles detected:")
print("Train:", train_file)
print("Validation:", val_file)
print("Test:", test_file)

if not all([train_file, val_file, test_file]):
    raise ValueError(
        "Could not identify all three files. "
        "Make sure their names contain train, val and test."
    )

# ============================================================
# 2. LOAD DATA
# ============================================================

train = pd.read_csv(train_file)
val = pd.read_csv(val_file)
test = pd.read_csv(test_file)

print("\nDataset shapes:")
print("Train      :", train.shape)
print("Validation :", val.shape)
print("Test       :", test.shape)

# ============================================================
# 3. VERIFY COLUMNS
# ============================================================

required = {"text", "label"}

for name, df in [
    ("Train", train),
    ("Validation", val),
    ("Test", test)
]:
    missing = required - set(df.columns)

    if missing:
        raise ValueError(
            f"{name} is missing columns: {missing}"
        )

# ============================================================
# 4. CLEAN TEXT
# ============================================================

def clean_text(text):

    text = str(text)

    # Remove URLs
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )

    # Normalize whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


train["clean_text"] = train["text"].apply(clean_text)
val["clean_text"] = val["text"].apply(clean_text)
test["clean_text"] = test["text"].apply(clean_text)

# ============================================================
# 5. SEPARATE X AND Y
# ============================================================

X_train = train["clean_text"]
y_train = train["label"]

X_val = val["clean_text"]
y_val = val["label"]

X_test = test["clean_text"]
y_test = test["label"]

# ============================================================
# 6. TF-IDF
# ============================================================

print("\nCreating TF-IDF features...")

vectorizer = TfidfVectorizer(
    lowercase=True,

    # word n-grams
    ngram_range=(1, 2),

    # ignore extremely rare words
    min_df=2,

    # ignore extremely common words
    max_df=0.95,

    # maximum vocabulary
    max_features=50000,

    sublinear_tf=True
)

X_train_tfidf = vectorizer.fit_transform(X_train)

X_val_tfidf = vectorizer.transform(X_val)

X_test_tfidf = vectorizer.transform(X_test)

print("\nTF-IDF shapes:")
print("Train:", X_train_tfidf.shape)
print("Val  :", X_val_tfidf.shape)
print("Test :", X_test_tfidf.shape)

print(
    "\nVocabulary size:",
    len(vectorizer.vocabulary_)
)

# ============================================================
# 7. TRAIN LOGISTIC REGRESSION
# ============================================================

print("\nTraining Logistic Regression...")

model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

model.fit(
    X_train_tfidf,
    y_train
)

print("Training complete.")

# ============================================================
# 8. VALIDATION PREDICTIONS
# ============================================================

val_pred = model.predict(X_val_tfidf)

# ============================================================
# 9. VALIDATION METRICS
# ============================================================

val_accuracy = accuracy_score(
    y_val,
    val_pred
)

val_macro_f1 = f1_score(
    y_val,
    val_pred,
    average="macro"
)

val_weighted_f1 = f1_score(
    y_val,
    val_pred,
    average="weighted"
)

precision, recall, f1, support = (
    precision_recall_fscore_support(
        y_val,
        val_pred,
        labels=[0, 1, 2],
        zero_division=0
    )
)

print("\n")
print("=" * 65)
print("VALIDATION RESULTS — TF-IDF + LOGISTIC REGRESSION")
print("=" * 65)

print(f"\nAccuracy      : {val_accuracy:.4f}")
print(f"Macro-F1 ⭐   : {val_macro_f1:.4f}")
print(f"Weighted-F1   : {val_weighted_f1:.4f}")

print("\nPer-class results:")

class_names = {
    0: "Against",
    1: "Favour",
    2: "None"
}

for i, label in enumerate([0, 1, 2]):

    print(
        f"{class_names[label]:8s} | "
        f"Precision: {precision[i]:.4f} | "
        f"Recall: {recall[i]:.4f} | "
        f"F1: {f1[i]:.4f} | "
        f"Support: {support[i]}"
    )

# ============================================================
# 10. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:\n")

print(
    classification_report(
        y_val,
        val_pred,
        labels=[0, 1, 2],
        target_names=[
            "Against",
            "Favour",
            "None"
        ],
        digits=4,
        zero_division=0
    )
)

# ============================================================
# 11. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_val,
    val_pred,
    labels=[0, 1, 2]
)

plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Against",
        "Favour",
        "None"
    ],
    yticklabels=[
        "Against",
        "Favour",
        "None"
    ]
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title(
    "Confusion Matrix — TF-IDF + Logistic Regression"
)

plt.tight_layout()

os.makedirs(
    "results/baseline",
    exist_ok=True
)

plt.savefig(
    "results/baseline/lr_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 12. SAVE VALIDATION RESULTS
# ============================================================

results = pd.DataFrame({
    "Model": [
        "TF-IDF + Logistic Regression"
    ],

    "Accuracy": [
        val_accuracy
    ],

    "Macro_F1": [
        val_macro_f1
    ],

    "Weighted_F1": [
        val_weighted_f1
    ],

    "Against_F1": [
        f1[0]
    ],

    "Favour_F1": [
        f1[1]
    ],

    "None_F1": [
        f1[2]
    ]
})

results.to_csv(
    "results/baseline/logistic_regression_validation.csv",
    index=False
)

print("\nValidation results saved.")

# ============================================================
# 13. SAVE PREDICTIONS
# ============================================================

val_predictions = val.copy()

val_predictions["predicted_label"] = val_pred

val_predictions["correct"] = (
    val_predictions["label"]
    ==
    val_predictions["predicted_label"]
)

val_predictions.to_csv(
    "results/baseline/logistic_regression_val_predictions.csv",
    index=False
)

# ============================================================
# 14. ERROR ANALYSIS PREVIEW
# ============================================================

errors = val_predictions[
    val_predictions["correct"] == False
]

print("\n")
print("=" * 65)
print("VALIDATION ERROR ANALYSIS PREVIEW")
print("=" * 65)

print(
    f"\nIncorrect predictions: {len(errors)}"
)

print(
    f"Validation accuracy: "
    f"{val_accuracy:.4f}"
)

print("\nSample errors:")

for _, row in errors.head(15).iterrows():

    print("\n----------------------------------------")

    print(
        "Actual    :",
        class_names.get(
            int(row["label"]),
            row["label"]
        )
    )

    print(
        "Predicted :",
        class_names.get(
            int(row["predicted_label"]),
            row["predicted_label"]
        )
    )

    print(
        "Text      :",
        row["text"]
    )

# ============================================================
# 15. IMPORTANT: DO NOT TOUCH TEST YET
# ============================================================

print("\n")
print("=" * 65)
print("IMPORTANT")
print("=" * 65)

print("""
We have NOT used the test set for model selection.

The test set will be used only after we decide
the final baseline configuration.

Next experiment:
TF-IDF + Linear SVM
""")

# ============================================================
# 16. DOWNLOAD RESULTS
# ============================================================

import shutil

shutil.make_archive(
    "baseline1_logistic_regression",
    "zip",
    "results/baseline"
)

files.download(
    "baseline1_logistic_regression.zip"
)

print("\nBaseline 1 complete.")
```

```python
# ============================================================
# EXPERIMENT 3: TF-IDF + CLASS-WEIGHTED LINEAR SVM
# CSET346 - Indic Climate Stance Challenge
# ============================================================

import pandas as pd
import numpy as np
import re
import os
import json

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------

TRAIN_PATH = "/content/train(1).csv"
VAL_PATH   = "/content/val(1).csv"
TEST_PATH  = "/content/test(1).csv"

train_df = pd.read_csv(TRAIN_PATH)
val_df   = pd.read_csv(VAL_PATH)
test_df  = pd.read_csv(TEST_PATH)

print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)

# ------------------------------------------------------------
# 2. BASIC TEXT PREPROCESSING
# Same preprocessing as Experiment 1 & 2
# ------------------------------------------------------------

def clean_text(text):
    text = str(text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", " ", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


X_train_text = train_df["text"].apply(clean_text)
X_val_text   = val_df["text"].apply(clean_text)
X_test_text  = test_df["text"].apply(clean_text)

y_train = train_df["label"]
y_val   = val_df["label"]
y_test  = test_df["label"]

# ------------------------------------------------------------
# 3. TF-IDF
# IMPORTANT: Fit ONLY on training data
# ------------------------------------------------------------

vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    max_features=50000,
    sublinear_tf=True
)

X_train = vectorizer.fit_transform(X_train_text)
X_val   = vectorizer.transform(X_val_text)
X_test  = vectorizer.transform(X_test_text)

print("\nTF-IDF shapes:")
print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)

# ------------------------------------------------------------
# 4. CLASS-WEIGHTED LINEAR SVM
# ------------------------------------------------------------

model = LinearSVC(
    C=1.0,
    class_weight="balanced",
    random_state=42
)

model.fit(X_train, y_train)

# ------------------------------------------------------------
# 5. VALIDATION PREDICTIONS
# ------------------------------------------------------------

val_pred = model.predict(X_val)

accuracy = accuracy_score(y_val, val_pred)
macro_f1 = f1_score(y_val, val_pred, average="macro")
weighted_f1 = f1_score(y_val, val_pred, average="weighted")

print("\n" + "=" * 60)
print("EXPERIMENT 3 RESULTS")
print("=" * 60)

print(f"Accuracy   : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Macro-F1   : {macro_f1:.4f} ({macro_f1*100:.2f}%)")
print(f"Weighted-F1: {weighted_f1:.4f} ({weighted_f1*100:.2f}%)")

# ------------------------------------------------------------
# 6. CLASS-WISE PERFORMANCE
# Labels:
# 0 = Against
# 1 = Favour
# 2 = None
# ------------------------------------------------------------

print("\nClassification Report:")
print(
    classification_report(
        y_val,
        val_pred,
        labels=[0, 1, 2],
        target_names=["Against", "Favour", "None"],
        digits=4,
        zero_division=0
    )
)

report = classification_report(
    y_val,
    val_pred,
    labels=[0, 1, 2],
    target_names=["Against", "Favour", "None"],
    output_dict=True,
    zero_division=0
)

print("\nClass F1 Scores:")
print(f"Against F1: {report['Against']['f1-score']:.4f}")
print(f"Favour  F1: {report['Favour']['f1-score']:.4f}")
print(f"None    F1: {report['None']['f1-score']:.4f}")

# ------------------------------------------------------------
# 7. CONFUSION MATRIX
# ------------------------------------------------------------

cm = confusion_matrix(y_val, val_pred, labels=[0, 1, 2])

print("\nConfusion Matrix:")
print(cm)

fig, ax = plt.subplots(figsize=(7, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Against", "Favour", "None"]
)

disp.plot(ax=ax, values_format="d")
plt.title("Experiment 3 - Class-Weighted SVM")
plt.tight_layout()

# ------------------------------------------------------------
# 8. SAVE RESULTS
# ------------------------------------------------------------

output_dir = "/content/experiment3_class_weighted_svm"
os.makedirs(output_dir, exist_ok=True)

# Validation predictions
predictions = pd.DataFrame({
    "text": val_df["text"],
    "true_label": y_val,
    "predicted_label": val_pred
})

predictions.to_csv(
    f"{output_dir}/class_weighted_svm_val_predictions.csv",
    index=False
)

# Metrics
metrics = pd.DataFrame([{
    "model": "TF-IDF + Class-Weighted Linear SVM",
    "accuracy": accuracy,
    "macro_f1": macro_f1,
    "weighted_f1": weighted_f1,
    "against_f1": report["Against"]["f1-score"],
    "favour_f1": report["Favour"]["f1-score"],
    "none_f1": report["None"]["f1-score"]
}])

metrics.to_csv(
    f"{output_dir}/class_weighted_svm_validation.csv",
    index=False
)

# Confusion matrix
fig.savefig(
    f"{output_dir}/class_weighted_svm_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

# ------------------------------------------------------------
# 9. DO NOT USE TEST LABELS FOR MODEL SELECTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("IMPORTANT")
print("=" * 60)
print("Test data was NOT used for model selection.")
print("Only validation results should be used to compare Experiments 1-3.")

# ------------------------------------------------------------
# 10. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("EXPERIMENT 3 COMPLETE")
print("=" * 60)

print(f"""
Model       : TF-IDF + Class-Weighted Linear SVM
C           : 1.0
Class Weight: balanced
N-grams     : (1, 2)
Max Features: 50,000

Validation Accuracy : {accuracy*100:.2f}%
Validation Macro-F1 : {macro_f1*100:.2f}%
Validation Weighted-F1: {weighted_f1*100:.2f}%

Against F1 : {report["Against"]["f1-score"]*100:.2f}%
Favour F1  : {report["Favour"]["f1-score"]*100:.2f}%
None F1    : {report["None"]["f1-score"]*100:.2f}%

Results saved in:
{output_dir}
""")
```

```python
# ============================================================
# EXPERIMENT 5 — XLM-RoBERTa
# Indic Climate Stance Challenge
# ============================================================

!pip -q install transformers accelerate sentencepiece

import os
import re
import shutil
import random
import numpy as np
import pandas as pd
import torch

from google.colab import files

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from torch.utils.data import Dataset

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    DataCollatorWithPadding,
    EarlyStoppingCallback
)

# ============================================================
# 1. CHECK GPU
# ============================================================

print("=" * 70)
print("DEVICE CHECK")
print("=" * 70)

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

if device == "cpu":
    print("\n⚠️ WARNING: GPU not detected.")
    print("Go to Runtime → Change runtime type → T4 GPU")

# ============================================================
# 2. UPLOAD DATASETS
# ============================================================

print("\nUpload:")
print("1. train(1).csv")
print("2. val(1).csv")
print("3. test(1).csv")

uploaded = files.upload()

train_file = next(
    f for f in uploaded
    if f.lower().startswith("train") and f.lower().endswith(".csv")
)

val_file = next(
    f for f in uploaded
    if f.lower().startswith("val") and f.lower().endswith(".csv")
)

test_file = next(
    f for f in uploaded
    if f.lower().startswith("test") and f.lower().endswith(".csv")
)

train_df = pd.read_csv(train_file)
val_df = pd.read_csv(val_file)
test_df = pd.read_csv(test_file)

print("\nDataset sizes:")
print("Train      :", train_df.shape)
print("Validation :", val_df.shape)
print("Test       :", test_df.shape)

# ============================================================
# 3. BASIC TEXT CLEANING
# ============================================================

def clean_text(text):
    text = str(text)

    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", " ", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text

train_df["text"] = train_df["text"].apply(clean_text)
val_df["text"] = val_df["text"].apply(clean_text)
test_df["text"] = test_df["text"].apply(clean_text)

# ============================================================
# 4. LABELS
#
# 0 = Against
# 1 = Favour
# 2 = None
# ============================================================

NUM_LABELS = 3

label_names = {
    0: "Against",
    1: "Favour",
    2: "None"
}

# ============================================================
# 5. XLM-R TOKENIZER
# ============================================================

MODEL_NAME = "xlm-roberta-base"

print("\nLoading tokenizer:", MODEL_NAME)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

MAX_LENGTH = 128

# ============================================================
# 6. DATASET CLASS
# ============================================================

class StanceDataset(Dataset):

    def __init__(self, dataframe, tokenizer, max_length=128):
        self.texts = dataframe["text"].tolist()
        self.labels = dataframe["label"].astype(int).tolist()
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):

        encoding = self.tokenizer(
            self.texts[idx],
            truncation=True,
            max_length=self.max_length
        )

        encoding["labels"] = self.labels[idx]

        return encoding

train_dataset = StanceDataset(
    train_df,
    tokenizer,
    MAX_LENGTH
)

val_dataset = StanceDataset(
    val_df,
    tokenizer,
    MAX_LENGTH
)

# ============================================================
# 7. MODEL
# ============================================================

print("\nLoading XLM-R model...")

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=NUM_LABELS,
    id2label=label_names,
    label2id={
        "Against": 0,
        "Favour": 1,
        "None": 2
    }
)

# ============================================================
# 8. CLASS WEIGHTS
# Same principle as our best classical baseline
# ============================================================

class_counts = (
    train_df["label"]
    .value_counts()
    .sort_index()
    .values
)

class_weights = len(train_df) / (
    NUM_LABELS * class_counts
)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
)

print("\nClass counts:")
print(class_counts)

print("\nClass weights:")
print(class_weights)

# ============================================================
# 9. CUSTOM TRAINER WITH CLASS-WEIGHTED LOSS
# ============================================================

class WeightedTrainer(Trainer):

    def compute_loss(
        self,
        model,
        inputs,
        return_outputs=False,
        num_items_in_batch=None
    ):

        labels = inputs.pop("labels")

        outputs = model(**inputs)

        logits = outputs.logits

        loss_function = torch.nn.CrossEntropyLoss(
            weight=class_weights.to(logits.device)
        )

        loss = loss_function(
            logits.view(-1, NUM_LABELS),
            labels.view(-1)
        )

        return (
            (loss, outputs)
            if return_outputs
            else loss
        )

# ============================================================
# 10. METRICS
# ============================================================

def compute_metrics(eval_pred):

    logits, labels = eval_pred

    predictions = np.argmax(
        logits,
        axis=-1
    )

    accuracy = accuracy_score(
        labels,
        predictions
    )

    macro_f1 = f1_score(
        labels,
        predictions,
        average="macro"
    )

    weighted_f1 = f1_score(
        labels,
        predictions,
        average="weighted"
    )

    return {
        "accuracy": accuracy,
        "macro_f1": macro_f1,
        "weighted_f1": weighted_f1
    }

# ============================================================
# 11. TRAINING SETTINGS
# ============================================================

OUTPUT_DIR = "/content/experiment5_xlmr"

training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,

    # Training
    num_train_epochs=3,

    per_device_train_batch_size=8,
    per_device_eval_batch_size=16,

    gradient_accumulation_steps=2,

    learning_rate=2e-5,
    weight_decay=0.01,

    warmup_ratio=0.1,

    # Evaluation
    eval_strategy="epoch",
    save_strategy="epoch",

    load_best_model_at_end=True,

    metric_for_best_model="macro_f1",
    greater_is_better=True,

    # Logging
    logging_strategy="steps",
    logging_steps=50,

    # Reproducibility
    seed=42,

    # Performance
    fp16=torch.cuda.is_available(),

    report_to="none",

    save_total_limit=2
)

# ============================================================
# 12. TRAINER
# ============================================================

trainer = WeightedTrainer(
    model=model,
    args=training_args,

    train_dataset=train_dataset,
    eval_dataset=val_dataset,

    tokenizer=tokenizer,

    data_collator=DataCollatorWithPadding(
        tokenizer=tokenizer
    ),

    compute_metrics=compute_metrics,

    callbacks=[
        EarlyStoppingCallback(
            early_stopping_patience=1
        )
    ]
)

# ============================================================
# 13. TRAIN
# ============================================================

print("\n" + "=" * 70)
print("STARTING XLM-R TRAINING")
print("=" * 70)

trainer.train()

# ============================================================
# 14. BEST VALIDATION RESULTS
# ============================================================

print("\n" + "=" * 70)
print("FINAL VALIDATION EVALUATION")
print("=" * 70)

evaluation = trainer.evaluate()

for key, value in evaluation.items():

    if isinstance(value, float):
        print(f"{key}: {value:.4f}")

# ============================================================
# 15. VALIDATION PREDICTIONS
# ============================================================

pred_output = trainer.predict(
    val_dataset
)

val_logits = pred_output.predictions

val_pred = np.argmax(
    val_logits,
    axis=-1
)

# ============================================================
# 16. DETAILED METRICS
# ============================================================

accuracy = accuracy_score(
    val_df["label"],
    val_pred
)

macro_f1 = f1_score(
    val_df["label"],
    val_pred,
    average="macro"
)

weighted_f1 = f1_score(
    val_df["label"],
    val_pred,
    average="weighted"
)

report = classification_report(
    val_df["label"],
    val_pred,
    labels=[0, 1, 2],
    target_names=[
        "Against",
        "Favour",
        "None"
    ],
    output_dict=True,
    zero_division=0
)

print("\n" + "=" * 70)
print("EXPERIMENT 5 RESULTS")
print("=" * 70)

print(
    f"Accuracy    : {accuracy:.4f} "
    f"({accuracy*100:.2f}%)"
)

print(
    f"Macro-F1    : {macro_f1:.4f} "
    f"({macro_f1*100:.2f}%)"
)

print(
    f"Weighted-F1 : {weighted_f1:.4f} "
    f"({weighted_f1*100:.2f}%)"
)

print("\nClass-wise F1:")

print(
    f"Against F1 : "
    f"{report['Against']['f1-score']:.4f} "
    f"({report['Against']['f1-score']*100:.2f}%)"
)

print(
    f"Favour F1  : "
    f"{report['Favour']['f1-score']:.4f} "
    f"({report['Favour']['f1-score']*100:.2f}%)"
)

print(
    f"None F1    : "
    f"{report['None']['f1-score']:.4f} "
    f"({report['None']['f1-score']*100:.2f}%)"
)

# ============================================================
# 17. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        val_df["label"],
        val_pred,
        labels=[0, 1, 2],
        target_names=[
            "Against",
            "Favour",
            "None"
        ],
        digits=4,
        zero_division=0
    )
)

# ============================================================
# 18. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    val_df["label"],
    val_pred,
    labels=[0, 1, 2]
)

print("Confusion Matrix:")
print(cm)

fig, ax = plt.subplots(
    figsize=(7, 6)
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Against",
        "Favour",
        "None"
    ]
)

disp.plot(
    ax=ax,
    values_format="d"
)

plt.title(
    "Experiment 5 — XLM-RoBERTa"
)

plt.tight_layout()

cm_path = (
    "/content/experiment5_xlmr/"
    "xlmr_confusion_matrix.png"
)

plt.savefig(
    cm_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 19. SAVE VALIDATION PREDICTIONS
# ============================================================

predictions_df = pd.DataFrame({
    "text": val_df["text"],
    "true_label": val_df["label"],
    "predicted_label": val_pred
})

predictions_df.to_csv(
    "/content/experiment5_xlmr/"
    "xlmr_validation_predictions.csv",
    index=False
)

# ============================================================
# 20. SAVE METRICS
# ============================================================

metrics_df = pd.DataFrame([{
    "model": "XLM-RoBERTa",
    "accuracy": accuracy,
    "macro_f1": macro_f1,
    "weighted_f1": weighted_f1,
    "against_f1": report["Against"]["f1-score"],
    "favour_f1": report["Favour"]["f1-score"],
    "none_f1": report["None"]["f1-score"]
}])

metrics_df.to_csv(
    "/content/experiment5_xlmr/"
    "xlmr_validation_metrics.csv",
    index=False
)

# ============================================================
# 21. SAVE MODEL + TOKENIZER
# ============================================================

BEST_MODEL_DIR = (
    "/content/experiment5_xlmr/"
    "best_model"
)

trainer.save_model(BEST_MODEL_DIR)
tokenizer.save_pretrained(BEST_MODEL_DIR)

# ============================================================
# 22. ZIP EVERYTHING
# ============================================================

zip_path = shutil.make_archive(
    "/content/experiment5_xlmr_results",
    "zip",
    "/content/experiment5_xlmr"
)

print("\n" + "=" * 70)
print("EXPERIMENT 5 COMPLETE")
print("=" * 70)

print(f"Validation Macro-F1: {macro_f1*100:.2f}%")

print("\nCurrent classical baseline:")
print("Class-Weighted TF-IDF + SVM = 46.97% Macro-F1")

if macro_f1 > 0.4697:
    print("\n🏆 XLM-R BEATS THE CURRENT BASELINE!")
else:
    print("\nXLM-R did not beat the current baseline yet.")

print("\nResults ZIP:")
print(zip_path)

print("\nIMPORTANT:")
print("Test labels were NOT used.")
print("Model selection is based ONLY on validation Macro-F1.")
```

```python
# ============================================================
# EXPERIMENT 6 — MuRIL
# Indic Climate Stance Challenge
# ============================================================

import os
import re
import shutil
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from google.colab import files

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from torch.utils.data import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    DataCollatorWithPadding
)

# ============================================================
# 1. GPU CHECK
# ============================================================

print("=" * 70)
print("DEVICE CHECK")
print("=" * 70)

print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("⚠️ GPU NOT DETECTED — use Colab T4 GPU.")

# ============================================================
# 2. LOAD DATA
# ============================================================

try:
    train_df
    val_df
    test_df
    print("\nUsing datasets already loaded in memory.")

except NameError:

    print("\nUpload:")
    print("1. train(1).csv")
    print("2. val(1).csv")
    print("3. test(1).csv")

    uploaded = files.upload()

    train_file = next(
        f for f in uploaded
        if f.lower().startswith("train") and f.lower().endswith(".csv")
    )

    val_file = next(
        f for f in uploaded
        if f.lower().startswith("val") and f.lower().endswith(".csv")
    )

    test_file = next(
        f for f in uploaded
        if f.lower().startswith("test") and f.lower().endswith(".csv")
    )

    train_df = pd.read_csv(train_file)
    val_df = pd.read_csv(val_file)
    test_df = pd.read_csv(test_file)

print("\nDataset sizes:")
print("Train      :", train_df.shape)
print("Validation :", val_df.shape)
print("Test       :", test_df.shape)

# ============================================================
# 3. TEXT CLEANING
# ============================================================

def clean_text(text):
    text = str(text)

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        " ",
        text
    )

    # Normalize whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text

train_df["text"] = train_df["text"].apply(clean_text)
val_df["text"] = val_df["text"].apply(clean_text)
test_df["text"] = test_df["text"].apply(clean_text)

# ============================================================
# 4. LABELS
# ============================================================

# 0 = Against
# 1 = Favour
# 2 = None

NUM_LABELS = 3

id2label = {
    0: "Against",
    1: "Favour",
    2: "None"
}

label2id = {
    "Against": 0,
    "Favour": 1,
    "None": 2
}

# ============================================================
# 5. MuRIL
# ============================================================

MODEL_NAME = "google/muril-base-cased"

print("\nLoading:", MODEL_NAME)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

MAX_LENGTH = 128

# ============================================================
# 6. DATASET
# ============================================================

class StanceDataset(Dataset):

    def __init__(
        self,
        dataframe,
        tokenizer,
        max_length=128
    ):

        self.texts = dataframe["text"].tolist()

        self.labels = (
            dataframe["label"]
            .astype(int)
            .tolist()
        )

        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):

        encoding = self.tokenizer(
            self.texts[idx],
            truncation=True,
            max_length=self.max_length
        )

        encoding["labels"] = self.labels[idx]

        return encoding

train_dataset = StanceDataset(
    train_df,
    tokenizer,
    MAX_LENGTH
)

val_dataset = StanceDataset(
    val_df,
    tokenizer,
    MAX_LENGTH
)

# ============================================================
# 7. MODEL
# ============================================================

print("\nLoading MuRIL model...")

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=NUM_LABELS,
    id2label=id2label,
    label2id=label2id
)

# ============================================================
# 8. CLASS WEIGHTS
# ============================================================

class_counts = (
    train_df["label"]
    .value_counts()
    .sort_index()
    .values
)

class_weights = (
    len(train_df)
    /
    (NUM_LABELS * class_counts)
)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float
)

print("\nClass counts:")
print(class_counts)

print("\nClass weights:")
print(class_weights)

# ============================================================
# 9. WEIGHTED TRAINER
# ============================================================

class WeightedTrainer(Trainer):

    def compute_loss(
        self,
        model,
        inputs,
        return_outputs=False,
        num_items_in_batch=None
    ):

        labels = inputs.pop("labels")

        outputs = model(**inputs)

        logits = outputs.logits

        loss_function = torch.nn.CrossEntropyLoss(
            weight=class_weights.to(
                logits.device
            )
        )

        loss = loss_function(
            logits.view(-1, NUM_LABELS),
            labels.view(-1)
        )

        if return_outputs:
            return loss, outputs

        return loss

# ============================================================
# 10. METRICS
# ============================================================

def compute_metrics(eval_pred):

    logits, labels = eval_pred

    predictions = np.argmax(
        logits,
        axis=-1
    )

    return {
        "accuracy": accuracy_score(
            labels,
            predictions
        ),

        "macro_f1": f1_score(
            labels,
            predictions,
            average="macro"
        ),

        "weighted_f1": f1_score(
            labels,
            predictions,
            average="weighted"
        )
    }

# ============================================================
# 11. TRAINING CONFIG
# Compatible with Transformers 5.x
# ============================================================

OUTPUT_DIR = "/content/experiment6_muril"

training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,

    num_train_epochs=3,

    per_device_train_batch_size=8,
    per_device_eval_batch_size=16,

    gradient_accumulation_steps=2,

    learning_rate=2e-5,

    weight_decay=0.01,

    warmup_steps=0,

    eval_strategy="epoch",
    save_strategy="epoch",

    load_best_model_at_end=True,

    metric_for_best_model="macro_f1",
    greater_is_better=True,

    logging_strategy="steps",
    logging_steps=50,

    seed=42,

    fp16=torch.cuda.is_available(),

    report_to="none",

    save_total_limit=2
)

# ============================================================
# 12. TRAINER
# ============================================================

trainer = WeightedTrainer(
    model=model,
    args=training_args,

    train_dataset=train_dataset,
    eval_dataset=val_dataset,

    processing_class=tokenizer,

    data_collator=DataCollatorWithPadding(
        tokenizer=tokenizer
    ),

    compute_metrics=compute_metrics
)

# ============================================================
# 13. TRAIN
# ============================================================

print("\n" + "=" * 70)
print("STARTING EXPERIMENT 6 — MuRIL")
print("=" * 70)

trainer.train()

# ============================================================
# 14. VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("MuRIL VALIDATION")
print("=" * 70)

evaluation = trainer.evaluate()

for key, value in evaluation.items():

    if isinstance(value, float):
        print(
            f"{key}: {value:.4f}"
        )

# ============================================================
# 15. PREDICTIONS
# ============================================================

prediction_output = trainer.predict(
    val_dataset
)

val_pred = np.argmax(
    prediction_output.predictions,
    axis=-1
)

y_true = val_df["label"].values

# ============================================================
# 16. FINAL METRICS
# ============================================================

accuracy = accuracy_score(
    y_true,
    val_pred
)

macro_f1 = f1_score(
    y_true,
    val_pred,
    average="macro"
)

weighted_f1 = f1_score(
    y_true,
    val_pred,
    average="weighted"
)

report = classification_report(
    y_true,
    val_pred,

    labels=[
        0,
        1,
        2
    ],

    target_names=[
        "Against",
        "Favour",
        "None"
    ],

    output_dict=True,

    zero_division=0
)

# ============================================================
# 17. PRINT RESULTS
# ============================================================

print("\n" + "=" * 70)
print("EXPERIMENT 6 — MuRIL RESULTS")
print("=" * 70)

print(
    f"Accuracy    : {accuracy:.4f} "
    f"({accuracy*100:.2f}%)"
)

print(
    f"Macro-F1    : {macro_f1:.4f} "
    f"({macro_f1*100:.2f}%)"
)

print(
    f"Weighted-F1 : {weighted_f1:.4f} "
    f"({weighted_f1*100:.2f}%)"
)

print("\nClass-wise F1:")

print(
    f"Against F1 : "
    f"{report['Against']['f1-score']:.4f} "
    f"({report['Against']['f1-score']*100:.2f}%)"
)

print(
    f"Favour F1  : "
    f"{report['Favour']['f1-score']:.4f} "
    f"({report['Favour']['f1-score']*100:.2f}%)"
)

print(
    f"None F1    : "
    f"{report['None']['f1-score']:.4f} "
    f"({report['None']['f1-score']*100:.2f}%)"
)

# ============================================================
# 18. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_true,
        val_pred,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Against",
            "Favour",
            "None"
        ],

        digits=4,

        zero_division=0
    )
)

# ============================================================
# 19. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    val_pred,

    labels=[
        0,
        1,
        2
    ]
)

print("Confusion Matrix:")
print(cm)

fig, ax = plt.subplots(
    figsize=(7, 6)
)

ConfusionMatrixDisplay(
    confusion_matrix=cm,

    display_labels=[
        "Against",
        "Favour",
        "None"
    ]
).plot(
    ax=ax,
    values_format="d"
)

plt.title(
    "Experiment 6 — MuRIL"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/muril_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 20. SAVE PREDICTIONS
# ============================================================

pd.DataFrame({
    "text": val_df["text"],
    "true_label": y_true,
    "predicted_label": val_pred
}).to_csv(
    f"{OUTPUT_DIR}/muril_validation_predictions.csv",
    index=False
)

# ============================================================
# 21. SAVE METRICS
# ============================================================

pd.DataFrame([{

    "model":
        "MuRIL + Class-Weighted Loss",

    "accuracy":
        accuracy,

    "macro_f1":
        macro_f1,

    "weighted_f1":
        weighted_f1,

    "against_f1":
        report["Against"]["f1-score"],

    "favour_f1":
        report["Favour"]["f1-score"],

    "none_f1":
        report["None"]["f1-score"]

}]).to_csv(
    f"{OUTPUT_DIR}/muril_validation_metrics.csv",
    index=False
)

# ============================================================
# 22. SAVE MODEL
# ============================================================

trainer.save_model(
    f"{OUTPUT_DIR}/best_model"
)

tokenizer.save_pretrained(
    f"{OUTPUT_DIR}/best_model"
)

# ============================================================
# 23. CREATE SMALL RESULTS ZIP
# Do NOT include model weights in the upload ZIP.
# ============================================================

small_zip = (
    "/content/"
    "experiment6_muril_results_only.zip"
)

import zipfile

with zipfile.ZipFile(
    small_zip,
    "w",
    zipfile.ZIP_DEFLATED
) as z:

    for filename in [
        "muril_validation_metrics.csv",
        "muril_validation_predictions.csv",
        "muril_confusion_matrix.png"
    ]:

        path = os.path.join(
            OUTPUT_DIR,
            filename
        )

        if os.path.exists(path):

            z.write(
                path,
                arcname=filename
            )

# ============================================================
# 24. FINAL COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("FINAL COMPARISON")
print("=" * 70)

print(
    f"Classical SVM : 46.97% Macro-F1"
)

print(
    f"XLM-R         : 52.97% Macro-F1"
)

print(
    f"MuRIL         : {macro_f1*100:.2f}% Macro-F1"
)

if macro_f1 > 0.5297:

    print(
        "\n🏆 MuRIL BEATS XLM-R!"
    )

elif macro_f1 > 0.4697:

    print(
        "\n✅ MuRIL beats the classical baseline "
        "but does not beat XLM-R."
    )

else:

    print(
        "\nMuRIL is below our current best models."
    )

print(
    "\nResults ZIP:"
)

print(small_zip)

print(
    "\n⚠️ TEST SET WAS NOT USED."
)
```
